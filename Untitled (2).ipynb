```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession.builder \
    .appName("Steam Reviews Analysis") \
    .getOrCreate()
```

```python
spark.sparkContext.master
```

```python
df = spark.read \
    .option("header", "true") \
    .option("multiLine", "true") \
    .option("escape", "\"") \
    .csv("file:///C:/Users/ALDA0056/steam_reviews.csv")
```

```python
df.count()
```

```python
df.printSchema()
```

```python
df.groupBy("recommended").count().show()
```

```python
df.groupBy("language") \
  .count() \
  .orderBy("count", ascending=False) \
  .show(10)
```

```python
df.groupBy("app_name") \
  .count() \
  .orderBy("count", ascending=False) \
  .show(10, truncate=False)
```

```python
df.groupBy("steam_purchase").count().show()
```

```python
df.groupBy("received_for_free").count().show()
```

```python
df.groupBy("steam_purchase", "recommended").count().show()

df.groupBy("received_for_free", "recommended").count().show()
```

```python
df2 = df.withColumn(
    "votes_helpful",
    col("votes_helpful").cast("int")
)
```

```python
df2.groupBy("recommended").avg("votes_helpful").show()
```

```python
df2 = df2.withColumn(
    "playtime_forever_num",
    col("`author.playtime_forever`").cast("double")
)
```

```python
df2.groupBy("recommended").avg("playtime_forever_num").show()
```
